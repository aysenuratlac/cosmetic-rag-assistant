```python
# Cell 2 - CSV oku + tabloyu göster
# Cell 1 - Imports
import os
import pandas as pd
import matplotlib.pyplot as plt
SUMMARY_PATH = r"C:\Users\aysenuratlac\Desktop\mat409-chatbot\reports\model_comparison_summary.csv"

if not os.path.exists(SUMMARY_PATH):
    raise FileNotFoundError(f"Bulunamadı: {SUMMARY_PATH}. Önce summary CSV üretildi mi?")

df = pd.read_csv(SUMMARY_PATH)

# Daha okunaklı görüntüleme için sütun sırası
ordered_cols = [
    "model",
    "n_questions",
    "context_recall_mean",
    "faithfulness_mean",
    "answer_relevancy_mean",
    "overall_mean",
    "context_recall_nan_count",
    "faithfulness_nan_count",
    "answer_relevancy_nan_count",
]
df = df[[c for c in ordered_cols if c in df.columns]]

df
```

```python
# Cell 3 - Metrik ortalamaları: yan yana bar grafik
metrics = ["context_recall_mean", "faithfulness_mean", "answer_relevancy_mean"]

plot_df = df.set_index("model")[metrics]

ax = plot_df.plot(kind="bar")  # matplotlib default colors
ax.set_title("Model Karşılaştırması - RAGAS Metrik Ortalamaları")
ax.set_ylabel("Skor (0-1)")
ax.set_xlabel("Model")
ax.set_ylim(0, 1)
ax.legend(title="Metrik")

plt.tight_layout()
plt.show()
```

```python
# Cell 4 - Overall mean: tek grafik
overall_df = df.set_index("model")[["overall_mean"]]

ax = overall_df.plot(kind="bar")
ax.set_title("Model Karşılaştırması - Overall Mean")
ax.set_ylabel("Skor (0-1)")
ax.set_xlabel("Model")
ax.set_ylim(0, 1)
ax.legend().remove()

# Bar üstüne değer yazdır
for p in ax.patches:
    ax.annotate(f"{p.get_height():.3f}", (p.get_x() + p.get_width()/2, p.get_height()),
                ha="center", va="bottom", fontsize=10)

plt.tight_layout()
plt.show()
```

```python
# Cell 6 - (Opsiyonel) Metrik ortalamaları: nokta grafiği (daha rapor tarzı)
melt = df.melt(id_vars=["model"], value_vars=metrics, var_name="metric", value_name="score")

fig, ax = plt.subplots()
for metric_name, g in melt.groupby("metric"):
    ax.plot(g["model"], g["score"], marker="o", label=metric_name)

ax.set_title("Model Karşılaştırması - Metrik Ortalamaları (Line)")
ax.set_ylabel("Skor (0-1)")
ax.set_xlabel("Model")
ax.legend(title="Metrik", loc="best")

plt.tight_layout()
plt.show()
```
